# Demo: LaTeX Math Rendering

## 1. Inline formula

Gradient Descent weight update: $w_{t+1} = w_t - \eta \frac{\partial L}{\partial w}$

## 2. Display-mode formula

$$\hat{y} = \sigma\left(\sum_{i=1}^{n} w_i x_i + b\right) = \frac{1}{1 + e^{-(\mathbf{w}^T \mathbf{x} + b)}}$$

## 3. Matrices

$$\mathbf{W} = \begin{bmatrix} w_{11} & w_{12} & w_{13} \\ w_{21} & w_{22} & w_{23} \end{bmatrix}, \quad \mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix}$$

## 4. Cross-Entropy Loss

$$L = -\frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$$

## 5. Bayes' Theorem

$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$$

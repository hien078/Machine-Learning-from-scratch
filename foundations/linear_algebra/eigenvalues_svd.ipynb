# Eigenvalues and Singular Value Decomposition

**Goal.** Connect symmetric eigendecomposition, SVD, rank, and low-rank approximation.


```python
import random

import matplotlib.pyplot as plt
import numpy as np

%load_ext autoreload
%autoreload 2

SEED = 42
random.seed(SEED)
rng = np.random.default_rng(SEED)
```

## WHAT

Every $X\in\mathbb R^{n\times d}$ has $X=U\Sigma V^\top$. The right singular vectors are eigenvectors of $X^\top X$, with eigenvalues $\sigma_i^2$.


```python
X = rng.normal(size=(8, 3))
U, singular_values, Vt = np.linalg.svd(X, full_matrices=False)
eigenvalues, eigenvectors = np.linalg.eigh(X.T @ X)
order = np.argsort(eigenvalues)[::-1]
np.testing.assert_allclose(eigenvalues[order], singular_values**2, atol=1e-12)
np.testing.assert_allclose(X, U @ np.diag(singular_values) @ Vt, atol=1e-12)

rank_one = singular_values[0] * np.outer(U[:, 0], Vt[0])
print(f"full Frobenius norm={np.linalg.norm(X):.4f}")
print(f"rank-one error={np.linalg.norm(X-rank_one):.4f}")
```

## CONNECTIONS

SVD gives stable least squares, PCA directions, numerical rank, and the best rank-$k$ approximation under the Frobenius norm.

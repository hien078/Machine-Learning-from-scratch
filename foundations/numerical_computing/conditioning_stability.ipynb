# Conditioning and Least-Squares Stability

**Goal.** Show that forming $X^\top X$ squares the condition number and can degrade a solve.


```python
import random

import matplotlib.pyplot as plt
import numpy as np

%load_ext autoreload
%autoreload 2

SEED = 42
random.seed(SEED)
rng = np.random.default_rng(SEED)
```

```python
x = np.linspace(0.99, 1.01, 50)
X = np.column_stack([np.ones_like(x), x, x**2, x**3])
true_w = np.array([1.0, -2.0, 3.0, -1.0])
y = X @ true_w + rng.normal(scale=1e-10, size=len(x))

condition_X = np.linalg.cond(X)
condition_gram = np.linalg.cond(X.T @ X)
stable = np.linalg.lstsq(X, y, rcond=None)[0]
normal_equation = np.linalg.solve(X.T @ X, X.T @ y)
print(f"cond(X)={condition_X:.3e}; cond(X.T @ X)={condition_gram:.3e}")
print(f"lstsq error={np.linalg.norm(stable-true_w):.3e}")
print(f"normal-equation error={np.linalg.norm(normal_equation-true_w):.3e}")
assert condition_gram > condition_X
```

## Result

The mathematical least-squares problem is unchanged, but the normal-equation algorithm operates on a more ill-conditioned matrix. Use QR/SVD-based `lstsq` by default.

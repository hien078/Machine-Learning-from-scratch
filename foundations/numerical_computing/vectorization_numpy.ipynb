# NumPy Vectorization

**Goal.** Replace sample loops with dimensionally explicit array operations and verify equality.


```python
import random
import time

import matplotlib.pyplot as plt
import numpy as np

%load_ext autoreload
%autoreload 2

SEED = 42
random.seed(SEED)
rng = np.random.default_rng(SEED)
```

```python
X = rng.normal(size=(2_000, 20))
w = rng.normal(size=20)

start = time.perf_counter()
loop = np.array([sum(row[j] * w[j] for j in range(len(w))) for row in X])
loop_seconds = time.perf_counter() - start

start = time.perf_counter()
vectorized = X @ w
vectorized_seconds = time.perf_counter() - start

np.testing.assert_allclose(loop, vectorized, atol=1e-12)
print(f"loop={loop_seconds:.6f}s; vectorized={vectorized_seconds:.6f}s")
```

## Result

Vectorization expresses the linear algebra directly and delegates loops to optimized compiled kernels. Timing is environment dependent; equality is the reproducible claim.

# Expectation, Variance, and Monte Carlo

**Goal.** Verify convergence of a sample mean and its $1/n$ variance scaling.


```python
import random

import matplotlib.pyplot as plt
import numpy as np

%load_ext autoreload
%autoreload 2

SEED = 42
random.seed(SEED)
rng = np.random.default_rng(SEED)
```

## WHAT

For independent variables with variance $\sigma^2$, $\operatorname{Var}(\bar{X}_n)=\sigma^2/n$. This describes estimator variability across repeated samples, not variation inside one sample.


```python
sample_sizes = np.array([5, 20, 100, 500])
empirical = []
for n in sample_sizes:
    means = rng.normal(loc=2.0, scale=3.0, size=(5_000, n)).mean(axis=1)
    empirical.append(np.var(means))
theory = 9.0 / sample_sizes
np.testing.assert_allclose(empirical, theory, rtol=0.12)
plt.loglog(sample_sizes, empirical, "o-", label="simulation")
plt.loglog(sample_sizes, theory, "--", label="9/n")
plt.xlabel("sample size n")
plt.ylabel("variance of sample mean")
plt.legend()
plt.show()
```

## CONNECTIONS

Sampling variability motivates confidence intervals, cross-validation uncertainty, and repeated-seed evaluation.

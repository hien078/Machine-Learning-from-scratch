# Probability Distributions

**Goal.** Connect Bernoulli, Gaussian, and Poisson assumptions to data types and likelihoods.


```python
import random

import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm, poisson

%load_ext autoreload
%autoreload 2

SEED = 42
random.seed(SEED)
rng = np.random.default_rng(SEED)
```

## WHAT

Bernoulli models a binary outcome, Poisson models nonnegative counts under a rate assumption, and Gaussian models a continuous variable with location and scale. Distribution choice is an assumption to test, not a default truth.


```python
figure, axes = plt.subplots(1, 2, figsize=(10, 4))
x = np.linspace(-4.0, 4.0, 300)
axes[0].plot(x, norm.pdf(x))
axes[0].set(title="Standard Gaussian density", xlabel="x", ylabel="density")
k = np.arange(0, 14)
axes[1].stem(k, poisson.pmf(k, mu=4.0))
axes[1].set(title="Poisson mass, rate 4", xlabel="count", ylabel="probability")
plt.show()
assert np.isclose(np.sum(poisson.pmf(np.arange(100), 4.0)), 1.0, atol=1e-12)
```

## CONNECTIONS

Gaussian noise yields squared error; Bernoulli likelihood yields binary cross-entropy; Gaussian class-conditionals yield Gaussian Naive Bayes.
